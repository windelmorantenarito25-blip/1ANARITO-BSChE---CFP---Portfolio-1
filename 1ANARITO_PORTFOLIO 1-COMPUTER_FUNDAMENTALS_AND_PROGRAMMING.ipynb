This program calculates the total budget for a visit to Quituinan Ranch in Camalig, Albay. It processes group size to determine costs for entrance fees, ATV rentals, and photos while demonstrating currency using floor division.

```python
# Program: Quituinan Ranch Trip Planner
# 1. DATA 
entry = 70         
atv = 200.0       
photo = 20

x = float(entry)
y = float(atv)
z = float(photo)

# 2. QUESTIONS
# Getting information from the user 
total_people = int(input('How many people are entering the ranch?'))
atv_people = int(input('How many people will ride the ATV?'))
photo_people = int(input('How many people want a souvenir photo?'))

# 3. CALCULATIONS

print('Number of Tourists:', total_people)
print('Number of People Availing ATV:', atv_people)
print('Number of People who Want to Take Pics:', photo_people)

x = float(entry)
y = float(atv)
z = float(photo)

print('Total cost for Entry only:', x * total_people, "PHP")
print('Total cost for ATV only:', y * atv_people, "PHP")
print('Total cost for Pics only:', z * photo_people, "PHP")

total_cost = (x * total_people) + (y * atv_people) + (z * photo_people)
print('The total cost for the group is:', (x * total_people) + (y * atv_people) + (z * photo_people), "PHP")
avg_per_person = total_cost / total_people
print('Average cost per person entering:', avg_per_person)

# Division using // rounds down to find bills
bills = total_cost // 500
print('Number of 500-peso bills:', bills)
# Finding the leftover change 
leftover = total_cost % 500
print('Loose change needed:', leftover)
```
